```python
import numpy as np
from matplotlib.typing import ColorType
import matplotlib.pyplot as plt
import random
```

```python
N = 200
x = np.random.uniform(0, 1, N)
eps = np.random.normal(-0.2, 0.1, N)
y = np.zeros(N)
for i in range(0,N):
  y[i] = (x[i])**2 + eps[i]
```

```python
plt.scatter(x, y, c = 'red',s=20)
plt.show()
```

```python
def L(w0,w1,x,y,N): # MSE
  sum = 0
  for j in range(0,N):
    sum += (y[j] - w1 * (x[j])**2 - w0)**2
  return sum / N
def grad_L(w0, w1, x, y, N):
  sum0 = 0
  sum1 = 0
  for j in range(0,N):
    sum0 += w1 * (x[j])**2 + w0 - y[j]
    sum1 += (w1 * (x[j])**2 + w0 - y[j]) * (x[j])**2
  sum0 *= 2/N
  sum1 *= 2/N
  return sum0, sum1
def stochastic_grad_L(w0, w1, x, y, batch, k):
  sum0 = 0
  sum1 = 0
  for j in range(k*batch, (k+1)*batch):
    sum0 += w1 * (x[j])**2 + w0 - y[j]
    sum1 += (w1 * (x[j])**2 + w0 - y[j]) * (x[j])**2
  sum0 *= 2/batch
  sum1 *= 2/batch
  return sum0, sum1
```

```python
# GD - Gradient descent
w0 = w1 = 0
alpha = 0.01
epoch = 10000
for i in range(0, epoch):
  w = grad_L(w0,w1,x,y,N)
  w0 = w0 - alpha * w[0]
  w1 = w1 - alpha * w[1]
w0, w1
```

```python
print("GD")
plt.scatter(x, y, c = 'red',s=20)
plt.plot([0, 1], [w0, w1+w0])
plt.show()
print("Loss = ", L(w0,w1,x,y,N))
```

```python
# SGD - Stochastic gradient descent
w0 = w1 = 0
alpha = 0.01
epoch = 1000
batch = 20
x0 = np.copy(x)
y0 = np.copy(y)
arr = np.column_stack((x0, y0))
for i in range(0, epoch):
  np.random.shuffle(arr)
  x0 = arr[:,0]
  y0 = arr[:, 1]
  for j in range(0, int(N / batch)):
    w = stochastic_grad_L(w0,w1,x0,y0,batch, j)
    w0 = w0 - alpha * w[0]
    w1 = w1 - alpha * w[1]
w0, w1
```

```python
print("SGD")
plt.scatter(x, y, c = 'red',s=20)
n = 20
h = 1/(n-1)
plt_x = np.zeros(n)
plt_y = np.zeros(n)
for i in range(0,n):
  plt_x[i] = h*i
  plt_y[i] = w1*(plt_x[i])**2 + w0
plt.plot(plt_x, plt_y)
plt.show()
print("Loss = ", L(w0,w1,x,y,N))
```
